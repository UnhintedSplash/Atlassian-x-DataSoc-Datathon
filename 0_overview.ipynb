# 0. Overview — The Blast Radius Blind Spot

Atlassian prioritises tickets by technical severity. That leaves a second
question unanswered: **how many collaborators may be affected when one
customer is blocked?**

**Business question:** Does current ticket priority recognise collaboration
reach, especially when highly collaborative customers are concentrated on
non-Free plans?

**This pipeline follows one evidence chain, one notebook per link:**

| Stage | Notebook | Question it answers |
|---|---|---|
| 1 | `1_data_foundation.ipynb` | Are the supplied fields and joins reliable? |
| 2 | `2_eda_population_and_queue.ipynb` | Who is represented, and how is the current queue composed? |
| 3 | `3_problem_discovery.ipynb` | What does the raw collaboration pattern look like? |
| 4 | `4_feature_engineering.ipynb` | How is collaboration reach turned into a usable feature? |
| 5 | `5_model_validation.ipynb` | Does collaboration relate to commercial-plan concentration — and is it missing from current priority? |
| 6 | `6_routing_simulation.ipynb` | Which tickets would a collaboration-aware routing rule actually change? |
| 7 | `7_pilot_readiness.ipynb` | What data would Atlassian need to validate this properly? |
| 8 | `8_recommendation.ipynb` | Given all of the above, what's the call? |

**How to read every stage:** findings are labelled as *Observed* (what the
data shows), *Business meaning* (how to interpret it), or *Limitation* (what
it does NOT prove). The data can establish relationships and workload; it
cannot prove revenue impact, SLA improvement, or causation — that
distinction is kept explicit throughout, not just here.

**Pipeline mechanics:** each stage loads whatever the previous stage saved
into `./artifacts/` and, if a later stage needs something new, saves its own
output back there. Run the notebooks in order the first time.
